### Solution for Task 2

```python
# SOLUTION
@mcp.tool()
async def batch_convert_documents_solution(file_paths: List[str], output_format: str = "markdown", output_dir: Optional[str] = None) -> str:
    """Batch convert multiple documents to a specified format.
    
    Args:
        file_paths: List of paths to documents
        output_format: Output format ('markdown' or 'json')
        output_dir: Directory to save converted files (optional)
    
    Returns:
        Summary of conversion results
    """
    try:
        # Validate output format
        if output_format not in ["markdown", "json"]:
            return "Error: output_format must be 'markdown' or 'json'"
        
        # Initialize converter
        converter = DocumentConverter()
        
        results = []
        
        for file_path in file_paths:
            try:
                if not os.path.exists(file_path):
                    results.append({
                        "file": file_path,
                        "status": "error",
                        "message": "File not found"
                    })
                    continue
                
                # Convert document
                result = converter.convert(file_path)
                
                # Export based on format
                if output_format == "markdown":
                    content = result.document.export_to_markdown()
                    ext = ".md"
                else:
                    content = json.dumps(result.document.export_to_dict(), indent=2)
                    ext = ".json"
                
                # Save if output directory specified
                output_path = None
                if output_dir:
                    os.makedirs(output_dir, exist_ok=True)
                    base_name = os.path.splitext(os.path.basename(file_path))[0]
                    output_path = os.path.join(output_dir, base_name + ext)
                    
                    with open(output_path, 'w', encoding='utf-8') as f:
                        f.write(content)
                
                results.append({
                    "file": file_path,
                    "status": "success",
                    "output_path": output_path,
                    "content_size": len(content)
                })
                
            except Exception as e:
                results.append({
                    "file": file_path,
                    "status": "error",
                    "message": str(e)
                })
        
        # Summary
        success_count = sum(1 for r in results if r["status"] == "success")
        error_count = sum(1 for r in results if r["status"] == "error")
        
        response = {
            "total_files": len(file_paths),
            "successful": success_count,
            "failed": error_count,
            "output_format": output_format,
            "output_directory": output_dir,
            "results": results
        }
        
        return json.dumps(response, indent=2)
        
    except Exception as e:
        return f"Error in batch conversion: {str(e)}"
```

### Solution for Task 3

```python
# SOLUTION
@mcp.tool()
async def chunk_document_for_rag_solution(file_path: str, chunk_size: int = 1000, chunk_overlap: int = 200, tokenizer: str = "cl100k_base") -> str:
    """Chunk a document for RAG applications using Docling's HybridChunker.
    
    Args:
        file_path: Path to the document
        chunk_size: Maximum tokens per chunk
        chunk_overlap: Token overlap between chunks
        tokenizer: Tokenizer to use (default: cl100k_base for OpenAI)
    
    Returns:
        Document chunks with metadata
    """
    try:
        from docling.chunking import HybridChunker
        
        # Check if file exists
        if not os.path.exists(file_path):
            return f"Error: File {file_path} does not exist"
        
        # Initialize converter
        converter = DocumentConverter()
        
        # Convert document
        result = converter.convert(file_path)
        
        # Initialize chunker
        chunker = HybridChunker(
            tokenizer=tokenizer,
            max_tokens=chunk_size,
            overlap=chunk_overlap
        )
        
        # Chunk the document
        chunks = list(chunker.chunk(result.document))
        
        # Format chunks
        formatted_chunks = []
        for i, chunk in enumerate(chunks):
            chunk_info = {
                "chunk_index": i,
                "text": chunk.text,
                "metadata": chunk.meta.export_json_dict() if hasattr(chunk.meta, 'export_json_dict') else {},
                "token_count": len(chunk.text.split())  # Approximate
            }
            formatted_chunks.append(chunk_info)
        
        response = {
            "status": "success",
            "file_path": file_path,
            "total_chunks": len(chunks),
            "chunk_size": chunk_size,
            "chunk_overlap": chunk_overlap,
            "tokenizer": tokenizer,
            "chunks": formatted_chunks
        }
        
        return json.dumps(response, indent=2)
        
    except Exception as e:
        return f"Error chunking {file_path}: {str(e)}"
```
